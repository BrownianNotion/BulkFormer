```python
data, _, _ = load_data("../UCLThesis/data/BIOAID_unlabelled_tpm_PC0.001_log2.parquet", return_df=True)
data
```

```python
bulkformer_gene_list = set(pd.read_csv("~/UCLThesis/data/bulkformer_gene_info.csv")["ensg_id"])
len(bulkformer_gene_list)
```

```python
labelled_df = pd.read_parquet("~/UCLThesis/data/BIOAID_UCL_Oxford_361_labelled.parquet")
labelled_df
```

```python
# Binary infection label: 1 for Viral or Bacterial & Viral, else 0
labelled_df['infection_status'] = labelled_df['micro_diagnosis'].isin(['Viral', 'Bacterial & Viral']).astype(int)

# Gene columns
gene_cols = [col for col in labelled_df.columns if col.startswith('ENSG') and col in bulkformer_gene_list]

# Efficiently replace log2(0.001) ≈ -9.965784 with -10 as mask token
labelled_df[gene_cols] = np.where(
    np.isclose(labelled_df[gene_cols].values, -9.965784, atol=0.2),
    -10,
    labelled_df[gene_cols].values
)

# Compute per-class Z-scores while treating -10 as NaN
ranked_genes = {}
for status in [0, 1]:
    subset = labelled_df[labelled_df['infection_status'] == status][gene_cols]
    masked = subset.replace(-10, np.nan)
    means = masked.mean()
    # stds = masked.std()
    
    # Remove small stds that blow up z-scores
    # valid_counts = masked.notna().sum()
    # stds[valid_counts < 20] = np.nan

    # z_scores = means / stds
    z_scores = means
    # Check for near-zero std
    # print("Genes with very small stds in class", status)
    # display(stds[stds < 1e-2].sort_values())  # REMOVE LATER
    ranked_genes[f'class_{status}'] = z_scores

# Compute and rank Z-score differences (class 1 - class 0)
zscore_diff = ranked_genes['class_1'] - ranked_genes['class_0']
zscore_diff = zscore_diff.sort_values(ascending=False)
zscore_diff
```

```python
MASK_PROB = 0.15
top_genes = int(MASK_PROB * 20010)
# just use 3000 honestly?
zscore_diff.index[:top_genes].to_series().to_csv("data/top_mean_diff_genes.csv", index=False)
```
